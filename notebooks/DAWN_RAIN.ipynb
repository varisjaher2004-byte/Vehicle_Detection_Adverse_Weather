```python
import os
import shutil

def backup_rain_training_data():
    # 1. Source Directory
    src_folder = r"C:\Users\Varis\runs\detect\DAWN_Rain_Project\train_yolov8l_rain-8"
    
    # 2. Destination Directory
    dest_folder = r"D:\DMSc_Dissertation_Project\datasets\DAWN\YOLO_RAIN\Backup_Staging_Runs"
    
    print("YOLOv8 Rain Run Complete Migration and Backup Tool")
    print("--------------------------------------------------\n")
    
    # Check if source directory exists
    if not os.path.exists(src_folder):
        print(f"Error: Source folder not found at: {src_folder}")
        return

    # Create destination directory if it does not exist
    if not os.path.exists(dest_folder):
        print(f"Creating directory: {dest_folder}")
        os.makedirs(dest_folder, exist_ok=True)
    
    print("Processing: Copying files...\n")
    
    # Loop through all files and subdirectories
    for root, dirs, files in os.walk(src_folder):
        rel_path = os.path.relpath(root, src_folder)
        target_path = os.path.join(dest_folder, rel_path) if rel_path != "." else dest_folder
        
        if not os.path.exists(target_path):
            os.makedirs(target_path, exist_ok=True)
            
        for file in files:
            source_file = os.path.join(root, file)
            destination_file = os.path.join(target_path, file)
            
            # Print status based on file type
            if file.endswith(('.png', '.jpg')):
                print(f"Copied Graph/Curve: {file}")
            elif file.endswith('.pt'):
                print(f"Copied Model Weight: {file}")
            elif file.endswith(('.csv', '.txt', '.yaml')):
                print(f"Copied Config/Log Matrix: {file}")
                
            shutil.copy2(source_file, destination_file)

    print("\nBackup complete. All files saved to destination.")

if __name__ == "__main__":
    backup_rain_training_data()
```

```python
import os
import xml.etree.ElementTree as ET

# PATHS

xml_folder = r"D:\DMSc_Dissertation_Project\datasets\DAWN\Rain\Rain_PASCAL_VOC"

label_folder = r"D:\DMSc_Dissertation_Project\datasets\DAWN\Rain\Rain_YOLO_FIXED"

os.makedirs(label_folder, exist_ok=True)

# CLASS MAPPING

class_map = {
    "person": 0,
    "rider": 1,
    "car": 2,
    "truck": 3,
    "bus": 4,
    "train": 5,
    "motorcycle": 6,
    "bicycle": 7
}

# CONVERT XML TO YOLO

count = 0

for file in os.listdir(xml_folder):

    if not file.endswith(".xml"):
        continue

    tree = ET.parse(os.path.join(xml_folder, file))
    root = tree.getroot()

    width = int(root.find("size/width").text)
    height = int(root.find("size/height").text)

    txt_file = os.path.join(
        label_folder,
        file.replace(".xml", ".txt")
    )

    with open(txt_file, "w") as f:

        for obj in root.findall("object"):

            name = obj.find("name").text.lower()

            if name not in class_map:
                continue

            cls = class_map[name]

            box = obj.find("bndbox")

            xmin = float(box.find("xmin").text)
            ymin = float(box.find("ymin").text)
            xmax = float(box.find("xmax").text)
            ymax = float(box.find("ymax").text)

            x = ((xmin + xmax) / 2) / width
            y = ((ymin + ymax) / 2) / height
            w = (xmax - xmin) / width
            h = (ymax - ymin) / height

            f.write(f"{cls} {x:.6f} {y:.6f} {w:.6f} {h:.6f}\n")

    count += 1

print(f"Conversion completed successfully.")
print(f"XML files converted: {count}")
print(f"Labels saved in:\n{label_folder}")
```
