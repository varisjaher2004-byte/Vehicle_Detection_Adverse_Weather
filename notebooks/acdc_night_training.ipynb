```python
import os
import shutil
import pandas as pd
import matplotlib.pyplot as plt

# 1. Configuration - Set these for each run (Night, Rain, Fog, Snow)
CONDITION = "night" 
PROJECT_DIR = r"C:\Users\Varis\runs\detect\ACDC_Night_Project\train_yolov8l_night-7"
BACKUP_DIR = fr"D:\DMSc_Dissertation_Project\datasets\ACDC\YOLO_{CONDITION.upper()}"

# Ensure backup directory exists
os.makedirs(BACKUP_DIR, exist_ok=True)

# 2. Save Dissertation Text Summary
summary_text = f"""
ACDC Dissertation Training Summary
Condition: {CONDITION.upper()}
Project Path: {PROJECT_DIR}
Status: Completed
Date: 2026-07-17
Notes: This backup contains weights, training logs, and performance visualizations for the ACDC {CONDITION} dataset evaluation.
"""
with open(os.path.join(BACKUP_DIR, "training_info.txt"), "w") as f:
    f.write(summary_text)

# 3. Copy Weights and Results
shutil.copy2(os.path.join(PROJECT_DIR, "weights", "best.pt"), BACKUP_DIR)
shutil.copy2(os.path.join(PROJECT_DIR, "results.csv"), BACKUP_DIR)

# 4. Professional Metrics Visualization
results_df = pd.read_csv(os.path.join(PROJECT_DIR, "results.csv"))
results_df.columns = results_df.columns.str.strip()

fig, ax = plt.subplots(1, 2, figsize=(15, 5))

# Plot 1: mAP metrics
ax[0].plot(results_df['epoch'], results_df['metrics/mAP50(B)'], label='mAP@50', color='blue')
ax[0].plot(results_df['epoch'], results_df['metrics/mAP50-95(B)'], label='mAP@50-95', color='green')
ax[0].set_title(f'mAP Progression - {CONDITION.upper()}')
ax[0].set_xlabel('Epochs')
ax[0].set_ylabel('mAP Score')
ax[0].legend()
ax[0].grid(True)

# Plot 2: Losses
ax[1].plot(results_df['epoch'], results_df['train/box_loss'], label='Box Loss', color='red')
ax[1].plot(results_df['epoch'], results_df['train/cls_loss'], label='Class Loss', color='orange')
ax[1].set_title(f'Training Loss - {CONDITION.upper()}')
ax[1].set_xlabel('Epochs')
ax[1].set_ylabel('Loss Value')
ax[1].legend()
ax[1].grid(True)

plt.tight_layout()
plt.savefig(os.path.join(BACKUP_DIR, "training_metrics_analysis.png"))
plt.show()

# 5. Final Summary CSV
summary_df = pd.DataFrame([{
    "Condition": CONDITION,
    "Max_mAP50": results_df['metrics/mAP50(B)'].max(),
    "Max_mAP50-95": results_df['metrics/mAP50-95(B)'].max()
}])
summary_df.to_csv(os.path.join(BACKUP_DIR, "performance_summary.csv"), index=False)

print(f"Professional backup and analysis for {CONDITION.upper()} completed successfully at {BACKUP_DIR}")
```
