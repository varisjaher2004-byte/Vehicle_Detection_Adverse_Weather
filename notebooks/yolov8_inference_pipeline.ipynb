```python
import os
from ultralytics import YOLO

# 1. Path Configuration
BASE_DIR = r"D:\DMSc_Dissertation_Project\datasets\Simulation_Project"
MODELS_DIR = os.path.join(BASE_DIR, "models")
INPUT_DIR = os.path.join(BASE_DIR, "input_data")
OUTPUT_DIR = os.path.join(BASE_DIR, "outputs")

os.makedirs(OUTPUT_DIR, exist_ok=True)

# 2. Define the correct mapping that your model SHOULD follow
# If your model's internal names are wrong, we use this to interpret the results
CORRECT_NAMES = {
    0: 'bicycle', 1: 'bus', 2: 'car', 3: 'motorcycle', 
    4: 'person', 5: 'truck', 6: 'traffic_light', 7: 'sign', 8: 'van'
}

print("Environment and Mapping initialized.")
```

```python
def run_batch_inference(mapping):
    for folder, model_file in mapping.items():
        print(f"\n--- Processing: {folder} | Model: {model_file} ---")
        
        model_path = os.path.join(MODELS_DIR, model_file)
        source_path = os.path.join(INPUT_DIR, folder)
        save_path = os.path.join(OUTPUT_DIR, folder)
        
        if not os.path.exists(model_path):
            print(f"Error: Model not found at {model_path}")
            continue
            
        model = YOLO(model_path)
        
        # Run inference
        # We perform the detection and then rely on our mapping for analysis
        results = model.predict(
            source=source_path,
            save=True,
            project=save_path,
            name="detection_results",
            conf=0.4,
            exist_ok=True
        )
        print(f"Results saved to: {save_path}")

print("Simulation engine updated.")
```

```python
experiment_mapping = {
    "acdc_entire_test": "acdc_entire_best.pt",
    "acdc_fog_test": "acdc_fog_best.pt",
    "acdc_night_test": "acdc_night_best.pt",
    "acdc_rain_test": "acdc_rain_best.pt",
    "acdc_snow_test": "acdc_snow_best.pt",
    "acdc+dawn_combined_test": "acdc+dawn_combined_best.pt",
    "dawn_entire_test": "dawn_entire_best.pt",
    "dawn_fog_test": "dawn_fog_best.pt",
    "dawn_rain_test": "dawn_rain_best.pt"
}

# Execute
run_batch_inference(experiment_mapping)
print("\n--- ALL SIMULATIONS COMPLETED ---")
```

```python
import os
from ultralytics import YOLO
import pandas as pd

# 1. Configuration
MODEL_PATH = r"D:\DMSc_Dissertation_Project\datasets\Simulation_Project\models\acdc_night_best.pt"
SOURCE_FOLDER = r"D:\DMSc_Dissertation_Project\datasets\Simulation_Project\input_data\acdc_night_test"
OUTPUT_FOLDER = r"D:\DMSc_Dissertation_Project\datasets\Simulation_Project\outputs\acdc_night_test"
OUTPUT_CSV = os.path.join(OUTPUT_FOLDER, "night_inference_results.csv")

# Ensure output directory exists
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# 2. Load the model
model = YOLO(MODEL_PATH)

# 3. Run Inference
print(f"Starting night-time inference for: {os.path.basename(MODEL_PATH)}")
# save=True will save the images with bounding boxes in the output folder
results = model.predict(source=SOURCE_FOLDER, conf=0.4, save=True, project=OUTPUT_FOLDER, name="visual_detections", verbose=False)

# 4. Extract data
data = []
for r in results:
    filename = os.path.basename(r.path)
    for box in r.boxes:
        cls_id = int(box.cls[0])
        conf = float(box.conf[0])
        data.append({'File': filename, 'Detected_ID': cls_id, 'Conf': conf})

# 5. Save to CSV
df = pd.DataFrame(data)
df.to_csv(OUTPUT_CSV, index=False)

print(f"Inference complete.")
print(f"Results saved to: {OUTPUT_CSV}")
print(f"Visualized images saved to: {os.path.join(OUTPUT_FOLDER, 'visual_detections')}")
print(f"Total detections found: {len(df)}")
```
