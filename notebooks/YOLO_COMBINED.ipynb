```python
import os

labels_dir = r"D:\DMSc_Dissertation_Project\datasets\YOLO_COMBINED\labels"

max_class = -1
bad_files = []

for root, _, files in os.walk(labels_dir):
    for file in files:
        if file.endswith(".txt"):
            path = os.path.join(root, file)

            with open(path, "r") as f:
                for line in f:
                    if line.strip() == "":
                        continue

                    cls = int(line.split()[0])

                    if cls > max_class:
                        max_class = cls

                    if cls < 0 or cls > 7:
                        bad_files.append((path, cls))

print("Maximum class ID:", max_class)
print("Invalid labels:", len(bad_files))

for item in bad_files[:20]:
    print(item)
```

```python
import os
import shutil
import pandas as pd

# 1. Configuration
PROJECT_DIR = r"C:\Users\Varis\runs\detect\YOLO_COMBINED\YOLOv8l_Final"
BACKUP_DIR = r"D:\DMSc_Dissertation_Project\datasets\YOLO_COMBINED"

# Create target directory
os.makedirs(BACKUP_DIR, exist_ok=True)

# 2. Define source paths
weights_dir = os.path.join(PROJECT_DIR, "weights")

# 3. List of files to archive
files_to_copy = [
    "args.yaml", "results.csv", "results.png", "confusion_matrix.png", 
    "confusion_matrix_normalized.png", "labels.jpg", "BoxF1_curve.png", 
    "BoxPR_curve.png", "BoxP_curve.png", "BoxR_curve.png"
]

print("Starting the professional backup process for YOLO_COMBINED...")

# Copy standard and curve files
for file in files_to_copy:
    src = os.path.join(PROJECT_DIR, file)
    if os.path.exists(src):
        shutil.copy2(src, BACKUP_DIR)
        print(f"Successfully archived: {file}")
    else:
        # Check in plots folder if not in root
        plots_src = os.path.join(PROJECT_DIR, "plots", file)
        if os.path.exists(plots_src):
            shutil.copy2(plots_src, BACKUP_DIR)
            print(f"Successfully archived: {file} (from plots)")

# Copy training/validation batch images
for file in os.listdir(PROJECT_DIR):
    if "batch" in file and file.endswith(".jpg"):
        shutil.copy2(os.path.join(PROJECT_DIR, file), BACKUP_DIR)
        print(f"Successfully archived: {file}")

# Copy weights
for weight_file in ["best.pt", "last.pt"]:
    src_weight = os.path.join(weights_dir, weight_file)
    if os.path.exists(src_weight):
        shutil.copy2(src_weight, BACKUP_DIR)
        print(f"Successfully archived: {weight_file}")

# 4. Create Performance Summary
results_df = pd.read_csv(os.path.join(PROJECT_DIR, "results.csv"))
results_df.columns = results_df.columns.str.strip()
summary_df = pd.DataFrame([{
    "Project": "YOLO_COMBINED_FINAL",
    "Max_mAP50": results_df['metrics/mAP50(B)'].max(),
    "Max_mAP50-95": results_df['metrics/mAP50-95(B)'].max()
}])
summary_df.to_csv(os.path.join(BACKUP_DIR, "performance_summary.csv"), index=False)

print(f"\nBackup complete! All files moved to: {BACKUP_DIR}")
```
